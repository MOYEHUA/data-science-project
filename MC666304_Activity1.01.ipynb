# 活动1.01 - 使用NumPy计算数据集的平均值、中位数、方差和标准差

```python
# 01 导入NumPy并设置别名np
import numpy as np
```

```python
# 02 使用genfromtxt加载数据集
dataset = np.genfromtxt('normal_distribution.csv', delimiter=',')
```

```python
# 03 打印数据集前两行的子集
dataset[0:2]
```

```python
# 04 计算第三行的平均值（索引2）
np.mean(dataset[2])
```

```python
# 05 获取每一行的最后一列
dataset[:, -1]
```

```python
# 06 获取每行前三个元素，前三列的子矩阵
dataset[:, 0:3]
```

```python
# 07 计算数据集最后一行的中位数
np.median(dataset[-1])
```

```python
# 08 获取最后三列
dataset[:, -3:]
```

```python
# 09 沿axis=1对每行求和
np.sum(dataset, axis=1)
```

```python
# 10 计算每列的方差，axis=0
np.var(dataset, axis=0)
```

```python
# 11 最后两行、前两列交集的方差
np.var(dataset[-2:, 0:2])
```

```python
# 12 计算整个数据集的标准差
np.std(dataset)
```
